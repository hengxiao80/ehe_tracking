```python
import numpy as np
import xarray as xr
from datetime import datetime, timedelta
# from scipy import stats
import h5py
import json
import pickle
import glob
```

```python
rd = 287.0
times = np.arange(7200, 14401, 30)
nt = len(times)
filelist = [f"OUT_3D/GOAMAZON_2pulse_256_zapped_{t:010d}.nc" for t in times]
for time in range(nt):
# for time in [120]:
    print(f"{time}: {filelist[time]})")
    infile = xr.open_dataset(filelist[time])
    t = infile.TABS
    qv = infile.QV
    qn = infile.QN
    p0 = infile.p
    pp = infile.PP
    tv = t*(1.0+0.61*(qv*1.0e-3)-(qn*1.0e-3))
    density = (pp + p0*100.0)/(rd*tv)
    density.to_dataset(name='density').to_netcdf(f'density/density_{time:08d}.nc')
```
