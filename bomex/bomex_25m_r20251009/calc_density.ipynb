```python
import numpy as np
import xarray as xr
from datetime import datetime, timedelta
# from scipy import stats
import h5py
import json
import pickle
import glob
```

```python
nt = 241
rd = 287.0
shift = 60
filelist = sorted(glob.glob('OUT_3D/*.nc'))
# print(filelist)
for time in range(nt):
# for time in [120]:
    # print(time)
    print(f'Processing time step {time}: {filelist[time+shift]}')
    infile = xr.open_dataset(filelist[time+shift])
    t = infile.TABS
    qv = infile.QV
    qn = infile.QN
    p0 = infile.p
    pp = infile.PP
    tv = t*(1.0+0.61*(qv*1.0e-3)-(qn*1.0e-3))
    density = (pp + p0*100.0)/(rd*tv)
    density.to_dataset(name='density').to_netcdf(f'density/density_{time:08d}.nc')
```
