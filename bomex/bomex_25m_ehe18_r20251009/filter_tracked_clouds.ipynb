Down select the cloud population for further analysis. 

```python
import json
```

```python
with open('hdf5/events.json', 'r') as f:
    all_clouds = json.load(f)
len(all_clouds)
```

```python
samplet = {k:v for (k,v) in all_clouds.items() if v['condense_time'] > 3}
len(samplet)
```

```python
samplet = {k:v for (k,v) in all_clouds.items() if v['condense_time'] > 2}
len(samplet)
```

```python
sample1 = {k:v for (k,v) in all_clouds.items() if v['condense_time'] > 1}
len(sample1)
```

```python
sample2 = {k:v for (k,v) in sample1.items() if v['cloud_volume'] > 7}
with open('large_clouds.json', 'w') as f:
    json.dump(sample2, f)
len(sample2)
```

```python
sample3 = {k:v for (k,v) in sample2.items() if not ("0" in v)}
```

```python
sample4 = {k:v for (k,v) in sample2.items() if ("240" in v or "0" in v)}
with open('interrupted_large_clouds.json', 'w') as f:
    json.dump(sample4, f)
len(sample4)
# for i, c in enumerate(sample4.keys()):
#     if int(c) < 176:
#         print(i, c)
```

```python
sample5 = {k:v for (k,v) in sample3.items() if not ("240" in v)}
with open('uninterrupted_large_clouds.json', 'w') as f:
    json.dump(sample5, f)
len(sample5)
```

```python
sample6 = {k:v for (k,v) in sample5.items() if v['core_volume'] > 7 and v['core_time'] > 1}
with open('uninterrupted_large_active_clouds.json', 'w') as f:
    json.dump(sample6, f)
len(sample6)
```
